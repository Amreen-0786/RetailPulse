```python
import pandas as pd
df=pd.read_csv('../data/retail_transactions.csv', parse_dates=['InvoiceDate'])
```

```python
df=df.drop_duplicates('TransactionID').dropna(subset=['CustomerID','InvoiceDate','Revenue'])
df.to_csv('../data/retail_transactions_clean.csv',index=False)
```

```python
snapshot=df.InvoiceDate.max()+pd.Timedelta(days=1)
rfm=df.groupby('CustomerID').agg(Recency=('InvoiceDate',lambda x:(snapshot-x.max()).days),Frequency=('TransactionID','nunique'),Monetary=('Revenue','sum'))
rfm.head()
```
